## Notes

```python
from pathlib import Path
import sys

PROJECT_ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'publication' / 'journal_article_1').exists()), None)
if PROJECT_ROOT is None:
    raise RuntimeError('Could not resolve project root from current working directory.')

SCRIPTS_DIR = PROJECT_ROOT / 'publication' / 'journal_article_1' / 'scripts'
if str(SCRIPTS_DIR) not in sys.path:
    sys.path.insert(0, str(SCRIPTS_DIR))

import notebook_artifact_build as nab
```

```python
section = '00'
print('=== Section 00: input validation and cleanup ===')
expected = nab.list_expected_artifacts(section)
print(f'Expected canonical artifacts in section {section}: {len(expected)}')
for p in expected:
    print(' -', p)

nab.ensure_required_inputs()
removed = nab.remove_stale_trial_artifacts()
print(f'Removed stale artifacts: {len(removed)}')
for p in removed:
    print(' -', p)
```

```python
print('=== Section 00: building tables and mapping ===')
outputs = nab.build_tables_and_mapping()
print(f'Generated outputs: {len(outputs)}')
for p in outputs:
    print('generated', p)
```

```python
print('=== Section 00: validation ===')
check = nab.check_artifacts('00')
display(check)

exists_count = int(check['exists'].sum())
missing_count = len(check) - exists_count
print(f'Validation summary: {exists_count}/{len(check)} artifacts exist; missing={missing_count}')

if missing_count:
    print('Missing artifacts:')
    for p in check.loc[~check['exists'], 'path']:
        print(' -', p)

nab.assert_artifacts('00')
print('Section 00 artifact check passed.')
```
